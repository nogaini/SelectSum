```python
def run_length_seg(arr):
    if not arr:
        return []
    groups = []

    prev_val = arr[0]
    group = [prev_val]
    for val in arr[1:]:
        if val == prev_val:
            group.append(val)
        else:
            groups.append(group)
            group = [val]
            prev_val = val
    groups.append(group)
    return groups
```

```python
run_length_seg([0, 0])
```
